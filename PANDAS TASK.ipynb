```python
# Task 1: Student Performance Dataset - Clean student academic data with multiple missing values. Fill marks columns using mean, Attendance using median, Section and City using mode.[cite: 2]

import pandas as pd
import numpy as np

data = {
    "Student": ["Asha", "Bala", "Charan", "Divya", "Eshan", "Farah"],
    "Maths": [90, np.nan, 78, 85, np.nan, 67],
    "Science": [88, 92, np.nan, 80, 75, np.nan],
    "English": [76, 81, np.nan, 79, 85, 88],
    "Attendance": [95, np.nan, 89, 91, 87, np.nan],
    "Section": ["A", "B", "A", np.nan, "B", "A"],
    "City": ["Chennai", "Salem", np.nan, "Madurai", "Salem", "Chennai"]
}

df = pd.DataFrame(data)


df["Maths"] = df["Maths"].fillna(df["Maths"].mean())
df["Science"] = df["Science"].fillna(df["Science"].mean())
df["English"] = df["English"].fillna(df["English"].mean())


df["Attendance"] = df["Attendance"].fillna(df["Attendance"].median())


df["Section"] = df["Section"].fillna(df["Section"].mode()[0])
df["City"] = df["City"].fillna(df["City"].mode()[0])

print(df)
```

```python
# Task 2: Employee Management Dataset - Clean employee management dataset. Fill Age using median, Salary and Bonus using mean, Experience using median, Department and City using mode.[cite: 2]

data = {
    "Employee": ["Ram", "Hari", "Kavi", "Meena", "John", "Sara"],
    "Age": [25, np.nan, 30, 28, np.nan, 35],
    "Salary": [35000, 42000, np.nan, 39000, 41000, np.nan],
    "Experience": [2, 5, np.nan, 4, 3, 7],
    "Bonus": [5000, np.nan, 7000, 6000, np.nan, 8000],
    "Department": ["HR", "IT", np.nan, "IT", "HR", "Finance"],
    "City": ["Chennai", np.nan, "Salem", "Coimbatore", "Salem", "Chennai"]
}

df = pd.DataFrame(data)

df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df["Experience"] = df["Experience"].fillna(df["Experience"].median())
df["Bonus"] = df["Bonus"].fillna(df["Bonus"].mean())

df["Department"] = df["Department"].fillna(df["Department"].mode()[0])
df["City"] = df["City"].fillna(df["City"].mode()[0])

print(df)
     
```

```python
# Task 3: Online Shopping Dataset - Handle customer shopping and payment data. Fill Amount and Discount using mean, Items and Delivery_Time using median, Payment Mode and City using mode.[cite: 2]

import pandas as pd
import numpy as np

data = {
    "Customer": ["Anu", "Raj", "Vijay", "Sara", "Kiran", "Deepa"],
    "Amount": [1200, np.nan, 1800, 1500, 2200, np.nan],
    "Items": [5, 3, np.nan, 4, 2, np.nan],
    "Discount": [100, 50, np.nan, 80, 60, np.nan],
    "Delivery_Time": [2, 4, np.nan, 3, 5, np.nan],
    "Payment Mode": ["UPI", "Card", np.nan, "Cash", "UPI", "Card"],
    "City": ["Salem", "Chennai", "Madurai", np.nan, "Salem", "Erode"]
}

df = pd.DataFrame(data)

for col in ["Amount", "Discount"]:
    df[col] = df[col].fillna(df[col].mean())

for col in ["Items", "Delivery_Time"]:
    df[col] = df[col].fillna(df[col].median())

for col in ["Payment Mode", "City"]:
    df[col] = df[col].fillna(df[col].mode()[0])

print(df)
```
