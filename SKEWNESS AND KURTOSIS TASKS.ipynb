```python
import matplotlib.pyplot as plt
import pandas as pd
from scipy.stats import stats 
from scipy.stats import kurtosis
```

```python
# ------------------------------------------------------------------------------
# Task 1: E-Commerce Order Value Analysis (Amazon / Flipkart)
# ------------------------------------------------------------------------------
orders = pd.Series([200, 250, 300, 350, 400, 450, 5000])


plt.hist(orders, bins=6, color='skyblue', edgecolor='black')
plt.title('Task 1: Order Value Distribution')
plt.xlabel('Order Value ($)')
plt.ylabel('Frequency')
plt.show()


skew1 = orders.skew()
skew_type1 = (
    'Positive Skew'
    if skew1 > 0
    else ('Negative Skew' if skew1 < 0 else 'Symmetric Distribution')
)

print(f'Task 1 - Skewness Value: {skew1:.4f}')
print(f'Task 1 - Distribution Type: {skew_type1}')

print('Conclusion: Few customers place huge orders.\n')
```

```python
ceo_data = pd.Series([25000, 26000, 27000, 28000, 29000, 30000, 500000])
plt.hist(ceo_data, bins=6, color='indianred', edgecolor='black')
plt.title('Kurtosis Task 1: CEO Salary Impact')
plt.xlabel('Salary ($)')
plt.ylabel('Frequency')
plt.show()
kurt1 = ceo_data.kurt()
kurt_type1 = (
    'Leptokurtic'
    if kurt1 > 0
    else ('Platykurtic' if kurt1 < 0  else 'Mesokurtic')
)
print(f'Kurtosis Value: {kurt1:.4f}')
print(f'Type: {kurt_type1}')
print('Reason: Extreme CEO salary outlier causes sharp peak/heavy tail.\n')
```

```python
stock_returns = pd.Series([-5, -4, -3, 0, 3, 4, 50])
plt.hist(stock_returns, bins=6, color='orchid', edgecolor='black')
plt.title('Kurtosis Task 2: Stock Market Returns')
plt.xlabel('Return (%)')
plt.ylabel('Frequency')
plt.show()
kurt2 = stock_returns.kurt()
kurt_type2 = (
    'Leptokurtic'
    if kurt2 > 0
    else ('Platykurtic' if kurt2 < 0 else 'Mesokurtic')
)
print(f'Kurtosis Value: {kurt2:.4f}')
print(f'Type: {kurt_type2}')
print('Reason: Highly volatile outlier return drives extreme peakedness.\n')
```

```python
heights = pd.Series([150, 152, 154, 156, 158, 160, 162])
plt.hist(heights, bins=5, color='mediumseagreen', edgecolor='black')
plt.title('Kurtosis Task 3: Student Heights')
plt.xlabel('Height (cm)')
plt.ylabel('Frequency')
plt.show()
kurt3 = heights.kurt()
kurt_type3 = (
    'Leptokurtic'
    if kurt3 > 0
    else ('Platykurtic' if kurt3 < 0 else 'Mesokurtic')
)
print(f'Kurtosis Value: {kurt3:.4f}')
print(f'Type: {kurt_type3}')
print('Reason: Uniform distribution creates light tails and a flat top.\n')
```

```python
quality_score = pd.Series([98, 99, 100, 100, 100, 101, 102])
plt.hist(quality_score, bins=5, color='cadetblue', edgecolor='black')
plt.title('Kurtosis Task 4: Manufacturing Quality')
plt.xlabel('Quality Index')
plt.ylabel('Frequency')
plt.show()
kurt4 = stats.kurtosis(quality_score)
if kurt4 > 0:
    kurt_type4 = 'Leptokurtic'
elif kurt4 < 0:
    kurt_type4 = 'Platykurtic'
else:
    kurt_type4 = 'Mesokurtic'
print(f'Kurtosis Value: {kurt4:.4f}')
print(f'Type: {kurt_type4}')
print('Reason: Flat/shallow distribution with low outlier intensity.\n')
```
