```python
yield_columns = [
    'RICE YIELD (Kg per ha)', 
    'WHEAT YIELD (Kg per ha)', 
    'MAIZE YIELD (Kg per ha)', 
    'CHICKPEA YIELD (Kg per ha)', 
    'GROUNDNUT YIELD (Kg per ha)', 
    'SUGARCANE YIELD (Kg per ha)', 
    'COTTON YIELD (Kg per ha)'
]
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df[yield_columns])
plt.title("Boxplot for 7 Crop Yield Metrics (Before Outlier Removal)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
rice = df['RICE YIELD (Kg per ha)']
Q1 = rice.quantile(0.25)
Q3 = rice.quantile(0.75)
IQR = Q3 - Q1  # Inter quantile ranging 
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
rice_outliers = df[(df['RICE YIELD (Kg per ha)'] < lower_bound) | (df['RICE YIELD (Kg per ha)'] > upper_bound)]
print("Rice Outliers Count:", len(rice_outliers))
```

```python
outlier_summary = []

for col in yield_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append([col, Q1, Q3, IQR, lower, upper, count])

outlier_df = pd.DataFrame(
    outlier_summary,
    columns=['Crop_Metric', 'Q1', 'Q3', 'IQR', 'Lower_Bound', 'Upper_Bound', 'Outlier_Count']
)

print(outlier_df)
```

```python
df_clean = df.copy()

for col in yield_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_clean = df_clean[
        (df_clean[col] >= lower) &
        (df_clean[col] <= upper)
    ]
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df[yield_columns])
plt.title("Agricultural Crop Yields - Before Outlier Removal")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df_clean[yield_columns])
plt.title("Agricultural Crop Yields - After Outlier Removal")
plt.xticks(rotation=45)
#plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df_clean['RICE YIELD (Kg per ha)'], kde=True)
plt.title("Rice Yield Distribution After Outlier Removal")
plt.xlabel("Rice Yield (Kg per ha)")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()
```
