**POISSON DISTRIBUTION**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson as p
lam=6
x=10 #x-> avg val
pos=p.pmf(x,lam)
print("Prob for getting visits for next 10 mins : ",pos)
xaxis=np.arange(0,21)
fit=p.pmf(xaxis,lam)
plt.bar(xaxis,fit,color="magenta")
plt.title("Poission Distribution")
plt.xlabel("No of calls per min")
plt.ylabel("Probability")
plt.show()
```

**CONFIDENCE INTERVAL**

```python
from scipy.stats import norm

# Given values
n = 50
mean = 165
sd = 10

# 95% confidence level -> Z value
z = 1.96

# Standard Error
se = sd / np.sqrt(n)

# Confidence Interval
lower = mean - z * se
upper = mean + z * se

print("95% Confidence Interval:")
print(lower, "to", upper)

# ---------------- GRAPH ----------------

x = np.linspace(150, 180, 500)
y = norm.pdf(x, mean, sd)

plt.plot(x, y)

# Confidence interval lines
plt.axvline(lower, linestyle='--')
plt.axvline(upper, linestyle='--')

plt.title("95% Confidence Interval")
plt.xlabel("Student Heights")
plt.ylabel("Density")

plt.show()
```

```python
x1=90
x2=110

mu=100 #mean
sig=15 #sd

prob_lower=norm.cdf(x1,mu,sig)
prob_greater=norm.cdf(x2,mu,sig)
print("Result will be :",round((prob_greater-prob_lower)*100,2),"%")

# To create random set of samples
X=np.random.normal(mu,sig,10000)

#histogram (Bell shaped curve)
plt.hist(X,bins=30,density=True,alpha=0.6)

# plot normal distribution curve
xmin,xmax=plt.xlim()
x=np.linspace(xmin,xmax,1000)
y=norm.pdf(x,mu,sig)
plt.plot(x,y,linewidth=2)

# shaded region
plt.fill_between(x,y,where=(x>=x1)&(x<=x2),alpha=0.2,color='red')

# graph
plt.title("Normal Distribution \n Bell-shaped Curve")
plt.xlabel("Height (in cm)")
plt.ylabel("Density")
plt.grid(True)
plt.show()
```

**UNIFORM DISTRIBUTION**

```python
start=1
end=100

lower = 30
upper = 60

# random uniform data
data=np.random.uniform(start,end,1000)

# Probability calculation
prob = (upper - lower) / (end - start)

# histogram
plt.hist(data,bins=30,density=True,edgecolor='black')

print("Probability between 30 and 60 :", prob)
print("Probability in percentage :", prob * 100, "%")



plt.axvline(30,color='red',linestyle='--')
plt.axvline(60,color='red',linestyle='--')

plt.title("Uniform Distribution")
plt.xlabel("Lucky Draw value")
plt.ylabel("Probability")
plt.grid(True)
# show plot
plt.show()
```
