```python
import re

# ==========================================
# Task 1: Extract Mobile Numbers
# Input: "Call me at 9876543210 or 9123456780"
# Goal: Extract all 10-digit mobile numbers using re.findall()
# ==========================================
text = "Call me at 9876543210 or 9123456780"
mobile_numbers = re.findall(r'\d{10}', text)
print(mobile_numbers)


# ==========================================
# Task 2: Extract Email IDs
# Input: "Emails: john@gmail.com, sara@yahoo.com"
# Goal: Extract all email addresses using \w+@\w+\.\w+
# ==========================================
text = "Emails: john@gmail.com, sara@yahoo.com"
emails = re.findall(r'\w+@\w+\.\w+', text)
print(emails)


# ==========================================
# Task 3: Find First Number
# Input: "Order ID: 45678"
# Goal: Check whether a number exists using re.search()
# ==========================================
text = "Order ID: 45678"
match = re.search(r'\d+', text)
if match:
    print(match.group())


# ==========================================
# Task 4: Check String Starts with Employee ID
# Input: "EMP12345 joined today"
# Goal: Verify whether string starts with "EMP" using ^ and re.match()
# ==========================================
text = "EMP12345 joined today"
match = re.match(r'^EMP', text)
if match:
    print("Match Found")


# ==========================================
# Task 5: Check PAN Number Format
# Input: "ABCDE1234F"
# Goal: Validate PAN format using [A-Z]{5}\d{4}[A-Z]
# ==========================================
text = "ABCDE1234F"
if re.search(r'^[A-Z]{5}\d{4}[A-Z]$', text):
    print("Valid PAN")


# ==========================================
# Task 6: Extract Product Codes
# Input: "Products: PRD123, PRD456, PRD789"
# Goal: Extract all product codes using [A-Z]{3}\d{3}
# ==========================================
text = "Products: PRD123, PRD456, PRD789"
product_codes = re.findall(r'[A-Z]{3}\d{3}', text)
print(product_codes)


# ==========================================
# Task 7: Remove Special Characters
# Input: "Data@Science#2026!"
# Goal: Remove special characters using re.sub() and \W
# ==========================================
text = "Data@Science#2026!"
clean_text = re.sub(r'\W', '', text)
print(clean_text)


# ==========================================
# Task 8: Extract Dates
# Input: "Meeting dates are 15-05-2026 and 20-06-2026"
# Goal: Extract dates using \d{2}-\d{2}-\d{4}
# ==========================================
text = "Meeting dates are 15-05-2026 and 20-06-2026"
dates = re.findall(r'\d{2}-\d{2}-\d{4}', text)
print(dates)


# ==========================================
# Task 9: Split Names
# Input: "John,Sara,David,Emma"
# Goal: Split names into list using re.split()
# ==========================================
text = "John,Sara,David,Emma"
names = re.split(r',', text)
print(names)


# ==========================================
# Task 10: Extract Hashtags
# Input: "Learning #Python #DataScience #AI"
# Goal: Extract hashtags using #\w+
# ==========================================
text = "Learning #Python #DataScience #AI"
hashtags = re.findall(r'#\w+', text)
print(hashtags)


# ==========================================
# Task 11: Find Usernames with Optional Underscore
# Input: "Users: john, sara_01, david"
# Goal: Extract usernames matching optional digits and underscores
# ==========================================
text = "Users: john, sara_01, david"
usernames = re.findall(r'\b[a-z]+(?:_\d+)?\b', text)
print(usernames)


# ==========================================
# Task 12: Extract Salary Values
# Input: "John earns ₹50000, Sara earns ₹65000"
# Goal: Extract salary amounts using \d+
# ==========================================
text = "John earns ₹50000, Sara earns ₹65000"
salaries = re.findall(r'\d+', text)
print(salaries)


# ==========================================
# Task 13: Find Department Names using OR Operator
# Input: "Departments: HR, IT, Finance, Sales"
# Goal: Extract only HR and IT using HR|IT
# ==========================================
text = "Departments: HR, IT, Finance, Sales"
departments = re.findall(r'HR|IT', text)
print(departments)


# ==========================================
# Task 14: Extract URLs
# Input: "Visit https://google.com and https://github.com"
# Goal: Extract website URLs using https://\w+\.\w+
# ==========================================
text = "Visit https://google.com and https://github.com"
urls = re.findall(r'https://\w+\.\w+', text)
print(urls)


# ==========================================
# Task 15: Find Iterative Matches with Position
# Input: "Employee IDs: EMP101 EMP102 EMP103"
# Goal: Extract IDs and their positions using re.finditer()
# ==========================================
text = "Employee IDs: EMP101 EMP102 EMP103"
for match in re.finditer(r'EMP\d+', text):
    print(f"{match.group()} found at position {match.start()}")
```
